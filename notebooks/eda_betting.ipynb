# EDA: betting market model

Tables from `analytics/betting/eda.py` over the fixtures of `tests/_fx_betting.py` (set `DATA_ROOT` to the collected odds on the Mac). **Fixture output, not a result**: odds history is a paid feed (The Odds API through `collect/extras.py --odds`) that has not been collected.

Output is tables and markdown only (no plotting library in the environment).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
from IPython.display import Markdown, display
from analytics.betting import eda

DATA_ROOT = None   # e.g. '/Users/.../data' on the Mac; None runs the fixtures
tables = eda.run_all(seed=0, data_root=DATA_ROOT, heldout_season='2024-25')
NOTE = '' if not tables['fixtures'] else ' (fixture output, not a result)'
display(Markdown('Held-out season: **%s**%s' % (tables['heldout_season'], NOTE)))
```

## Join coverage

Odds to games on the team nickname key and a game date within one day of the UTC tip minus 8 hours; duplicate closing snapshots resolved to the one nearest tip and counted.

```python
display(Markdown('### join coverage' + NOTE))
display(Markdown(eda.render_markdown(tables['join_coverage'])))
```

## Spread and margin distributions

`home_margin_minus_market` is the home margin plus the closing home spread: the line's error, centred near zero when the market is efficient.

```python
display(Markdown('### closing spread' + NOTE))
display(Markdown(eda.render_markdown(tables['spread_distribution'], nd=2)))
display(Markdown('### final margin' + NOTE))
display(Markdown(eda.render_markdown(tables['margin_distribution'], nd=2)))
```

## Cover rate by spread bin

A fair closing line covers about half the time in every bin; a bin far from 0.5 with a large n is where a model could add value.

```python
display(Markdown('### cover rate by spread bin' + NOTE))
display(Markdown(eda.render_markdown(tables['cover_rate_by_spread_bin'], nd=3)))
```

## Shuffled-label check

Out-of-fold cover log loss of each candidate on the true labels and on labels shuffled within season (margin, spread and cover shuffled together). On shuffled labels every model should sit near the market's 0.6931 or worse; a model that still beats the market on shuffled labels is leaking the outcome through a feature.

```python
display(Markdown('### shuffled-label check' + NOTE))
display(Markdown(eda.render_markdown(tables['shuffled_label_check'])))
```
