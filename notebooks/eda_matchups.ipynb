# EDA: individual matchups (analytics/matchups)

Tables from `analytics.matchups.eda` on the synthetic season of `tests/_fx_matchups.py` (or `analytics.fixtures` when present). **Every number in this notebook is fixture output, not a result.** On the Mac, replace `fixture_tables` with the tables of a real DATA_ROOT run (`analytics.matchups.validate.load_games` then `analytics.matchups.features.run_games`).

Output is tables and markdown only: no plotting library is used.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().resolve().parent))

from analytics.matchups import eda

tables = eda.fixture_tables(seed=0, n_games=2)
print('matchup rows', tables['n_rows'], 'games', tables['n_games'], '(fixture output, not a result)')
```

## Distribution of partial possessions

Share of a possession's sampled frames a defender was assigned to one offensive player. A tall last bin means assignments mostly hold for the whole possession; mass in the low bins comes from switches and short help assignments.

```python
print(tables['partial_possession'].to_markdown(index=False))
```

## Switches per defender-possession

The number of sampled frames at which a defender's assignment changed from the previous sampled frame.

```python
print(tables['switches'].to_markdown(index=False))
```

## Defended FG% by distance to the assigned defender

FG% of attempts by the distance between the shooter and its assigned defender at the sampled frame at or before the release. `mean_xfg` is the location-only probability (the default when no xFG predictions are joined here).

```python
print(tables['defended_fg'].to_markdown(index=False))
```

## Shuffled-assignment check

The FG% gap (within 4 ft minus beyond 4 ft) under the real assignment and under shuffles that give each attempt the distance of a random defender present at that frame. When the assignment carries information the real gap sits outside the shuffled band.

```python
print(tables['shuffled'].to_markdown(index=False))
```
